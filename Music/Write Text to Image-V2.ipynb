```python
from PIL import Image, ImageDraw, ImageFont

# Use a TrueType font with a larger size
font = ImageFont.truetype("arial.ttf", 160)

# Load your image
image_path = "Folder.png"  # Change if needed
img = Image.open(image_path)

# Create a drawing context
draw = ImageDraw.Draw(img)

# Define text
song_title  = "Time of Our Lives"
artist_name = "Zho Lei"

# Choose font and color
font = ImageFont.truetype("arial.ttf", 40)
text_color = (255, 255, 255)  # white

# Get image dimensions
width, height = img.size

# Position for middle right
x_position = int(width * 0.40)
y_position = int(height * 0.70)

# Draw text
draw.text((x_position, y_position), song_title, font=font, fill=text_color)
draw.text((x_position, y_position + 40), artist_name, font=font, fill=text_color)

# Convert to RGB before saving as JPEG
if img.mode == 'RGBA':
    img = img.convert('RGB')

# Save result
img.save("Folder.jpg")
print("Saved as Folder.jpg")
```
