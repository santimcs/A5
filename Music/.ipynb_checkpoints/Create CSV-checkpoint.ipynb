```python
import pandas as pd
from sqlalchemy import create_engine, text
    
engine = create_engine('mysql+pymysql://root@localhost:3306/music_development')
music = engine.connect()
data_path = '../data/'
```

```python
sql = '''
SELECT * FROM songs
WHERE rank BETWEEN 1060 AND 1069
'''
songs = pd.read_sql(sql, music)
songs
```

```python
songs.dtypes
```

```python
sql = '''
SELECT * FROM songs
WHERE name LIKE "yesterday"
'''
songs = pd.read_sql(sql, engine)
songs 
```

```python
sql = '''
SELECT * FROM songs
WHERE name LIKE "yesterday"
'''
songs = pd.read_sql(sql, engine)
song = songs.iloc[0]
songId = song['id']
songId
```

```python
# Method 2: Direct string printing with proper formatting
sql = f"""
SELECT id, song_id, content 
FROM lyrics 
WHERE song_id = {songId}
"""
lyric = pd.read_sql(sql, engine)
lyric
```

```python
lyric.dtypes
```

```python
sqlUpd = text("""UPDATE lyrics SET song_id = 531 WHERE id = 450""")
print(sqlUpd)
```

```python
rp = music.execute(sqlUpd, {'song_id': songId})

# Print the number of rows affected
print("Rows updated:", rp.rowcount)
```
